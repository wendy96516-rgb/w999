## 題目說明
整理「桌遊體驗」與「電影欣賞」兩場活動的原始報名名單。先進行資料清理（統一轉小寫、移除空白），再使用集合進行去重與交叉比對，最後利用字典儲存摘要資訊，並以 f-string 格式化輸出完整的報名、人數與費用報表。

```markdown
## 測試一：原始資料
```

````python
# 1. 原始資料定義
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 2. 讀取活動名稱與資料清理 (轉小寫、移除空白、以逗號分割)
board_name = activity_names[0]
movie_name = activity_names[1]

# 清理桌遊名單
board_clean = board_raw.replace(" ", "").lower()
board_list = board_clean.split(",")

# 清理電影名單
movie_clean = movie_raw.replace(" ", "").lower()
movie_list = movie_clean.split(",")

# 3. 集合去重與比較名單
board_set = set(board_list)
movie_set = set(movie_list)

common_participants = board_set & movie_set             # 共同參加者 (交集)
all_participants = board_set | movie_set                # 全部參加者 (聯集)
board_only = board_set - movie_set                      # 只參加桌遊者 (差集)
movie_only = movie_set - board_set                      # 只參加電影者 (差集)

# 轉成排序好的 list 以便顯示
board_unique_sorted = sorted(list(board_set))
movie_unique_sorted = sorted(list(movie_set))
common_sorted = sorted(list(common_participants))
all_sorted = sorted(list(all_participants))
board_only_sorted = sorted(list(board_only))
movie_only_sorted = sorted(list(movie_only))

# 4. 通知名單預覽 (前兩位)
preview_list = all_sorted[:2]

# 5. 建立 summary 字典與計算
summary = {
    "board_count": len(board_set),
    "movie_count": len(movie_set),
    "common_count": len(common_participants),
    "unique_count": len(all_participants),
    "total_registrations": len(board_set) + len(movie_set),
    "board_fee": len(board_set) * unit_fees[board_name],
    "movie_fee": len(movie_set) * unit_fees[movie_name]
}
summary["total_fee"] = summary["board_fee"] + summary["movie_fee"]

# 6. 格式化輸出
print("HW09 活動報名摘要 (測試一)")
print(f"活動：{board_name}，每人 {unit_fees[board_name]:.2f} 元")
print(f"活動：{movie_name}，每人 {unit_fees[movie_name]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_unique_sorted}")
print(f"電影去重名單：{movie_unique_sorted}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{common_sorted}")
print(f"共同人數：{summary['common_count']} 人")
print(f"只參加桌遊者：{board_only_sorted}")
print(f"只參加電影者：{movie_only_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['unique_count']} 人")
print(f"報名人次：{summary['total_registrations']} 人次")
print(f"通知名單前兩位：{preview_list}")
print(f"桌遊應收：{summary['board_fee']:.2f} 元")
print(f"電影應收：{summary['movie_fee']:.2f} 元")
print(f"總應收：{summary['total_fee']:.2f} 元")
````
